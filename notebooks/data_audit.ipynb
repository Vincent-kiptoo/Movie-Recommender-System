### runtime and budget should be converted from Varchar to interger/ numeric

```python
def convert_data_type(imdb_data: pd.DataFrame)-> pd.DataFrame:
    runtime = imdb_data["runtime"].astype("string").str.strip()
    imdb_data["runtime"] = runtime.mask(runtime.eq("NA"), pd.NA)
    imdb_data["runtime"] = pd.to_numeric(imdb_data["runtime"], errors="coerce",)
    imdb_data["budget"] = (imdb_data["budget"].astype("string").str.strip().str.replace("$", "", regex=False).str.replace(",", "", regex=False))

    imdb_data["budget"] = pd.to_numeric(imdb_data["budget"], errors="coerce")
    return imdb_data

imdb_data = convert_data_type(imdb_data)
```

```python
%%sql
-- Check missing plot keywords
SELECT
    COUNT(*) AS total_rows,
    COUNT(*) FILTER (WHERE plot_keywords IS NULL) AS missing_values,
    ROUND(100.0 * COUNT(*) FILTER (WHERE plot_keywords IS NULL) / COUNT(*), 2) AS missing_percentage
FROM imdb_data;
```

```python
%%sql 
SELECT * FROM imdb_data WHERE plot_keywords IS NULL
```

```python
%%sql

SELECT
    COUNT(*) FILTER (
        WHERE plot_keywords IS NULL
    ) AS null_values,

    COUNT(*) FILTER (
        WHERE TRIM(plot_keywords) = ''
    ) AS empty_strings

FROM imdb_data;
```

```python
%%sql

SELECT
    COUNT(*) AS joined_rows,
    COUNT(DISTINCT m.movieId) AS unique_movies,
    COUNT(*) FILTER (
        WHERE i.plot_keywords IS NULL
    ) AS missing_keywords,
    ROUND(
        100.0 * COUNT(*) FILTER (
            WHERE i.plot_keywords IS NULL
        ) / COUNT(*),
        2
    ) AS missing_percentage
FROM movies AS m
INNER JOIN imdb_data AS i
    ON m.movieId = i.movieId;
```

```python
%%sql
-- Check missing tmbid values
SELECT
    COUNT(*) AS total_rows,
    COUNT(*) FILTER (
        WHERE tmdbId IS NULL
    ) AS missing_values,
    ROUND(
        100.0 * COUNT(*) FILTER (
            WHERE tmdbId IS NULL
        ) / COUNT(*),
        2
    ) AS missing_percentage
FROM links;
```
