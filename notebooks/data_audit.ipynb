```python
%%sql
-- Check missing plot keywords
SELECT
    COUNT(*) AS total_rows,
    COUNT(*) FILTER (WHERE plot_keywords IS NULL) AS missing_values,
    ROUND(100.0 * COUNT(*) FILTER (WHERE plot_keywords IS NULL) / COUNT(*), 2) AS missing_percentage
FROM imdb_data;
```

```python
%%sql 
SELECT * FROM imdb_data WHERE plot_keywords IS NULL
```

```python
%%sql
--Checks the total number of null values and empty strings if it exists in imdb_data 

SELECT
    COUNT(*) FILTER (WHERE plot_keywords IS NULL) AS null_values,
    COUNT(*) FILTER (WHERE TRIM(plot_keywords) = '') AS empty_strings

FROM imdb_data;
```

```python
%%sql
-- Check the number of missing keywords after  inner joining imdb_data and movie data
SELECT
    COUNT(*) AS joined_rows,
    COUNT(DISTINCT m.movieId) AS unique_movies,
    COUNT(*) FILTER (WHERE i.plot_keywords IS NULL) AS missing_keywords,
    ROUND(100.0 * COUNT(*) FILTER (WHERE i.plot_keywords IS NULL) / COUNT(*), 2) AS missing_percentage
FROM movies AS m
INNER JOIN imdb_data AS i
    ON m.movieId = i.movieId;
```

# Data Audit: Findings and Planned Actions

## 1. Findings

The data audit identified missing values in only two datasets:

- **`imdb_data`:** The `plot_keywords` feature contains 1,250 missing values, representing 4.58% of the dataset.
- **`links`:** The `tmdbId` column contains 107 missing values, representing 0.17% of the dataset.

No immediate evidence from the audit justifies automatic row deletion or arbitrary imputation for either dataset. The appropriate handling strategy will depend on the relevance of the affected features to the recommendation models.

Additionally, the `budget` and `runtime` columns in `imdb_data` are stored as `VARCHAR` rather than numeric data types. This limits their direct use in numerical analysis and feature engineering.

## 2. Planned Actions

1. **Investigate missing values:** Examine the distribution and characteristics of missing `plot_keywords` and `tmdbId` values to determine their potential impact on downstream analysis and recommendation quality.

2. **Determine missing-value handling strategies:** Evaluate whether to retain missing values, apply appropriate imputation, or exclude affected records or features based on their relevance to the modelling objectives.

3. **Standardize data types:** Clean and convert `budget` and `runtime` from `VARCHAR` to suitable numeric types, handling currency symbols, separators, non-numeric values, and missing-value representations where applicable.

4. **Validate data quality:** Check the converted columns for invalid values, unexpected ranges, and conversion failures to preserve data integrity.

5. **Proceed with exploratory data analysis:** Examine feature distributions, relationships, outliers, and other relevant patterns to inform subsequent feature engineering and model development.

6. **Prepare modelling datasets:** Use the audit findings and subsequent analyses to develop consistent, model-ready datasets for the content-based, collaborative, and hybrid recommendation approaches.
